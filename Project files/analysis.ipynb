## Retail Sales Data Analysis

## Objective
The objective of this project is to analyze retail sales data to identify trends, top-performing products, seasonal patterns, and regional performance, and to generate actionable business insights.

## Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load Dataset

```python
data = pd.read_csv("C:/Sample - Superstore.csv",engine="python",encoding="latin-1")
data.head()
```

## Data Understanding

```python
data.shape
```

```python
data.info()
```

```python
data.isnull().sum()
```

## Data Cleaning

```python
data["Order Date"] = pd.to_datetime(data["Order Date"])
data["Month"] = data["Order Date"].dt.to_period("M")
```

Converted date format and created Month column for trend analysis.

```python
data.info()
```

## Key Performance Indicators (KPIs)

```python
total_revenue = data["Sales"].sum()
print("Total Revenue:", total_revenue)
average_order_value = data["Sales"].mean()
print("Average Order Value:", average_order_value)
top_region = data.groupby("Region")["Sales"].sum().idxmax()
print("Top Region:", top_region)
```

- Total revenue shows overall business performance  
- Average order value indicates customer spending behavior  
- Top region highlights highest contributing market  

## Visualization

## Sales Over Time

```python
monthly_sales = data.groupby("Month")["Sales"].sum()
monthly_sales.plot(figsize=(10,5))
plt.title("Sales Over Time")
plt.show()
```

Sales show an overall trend with fluctuations, indicating seasonal demand patterns.

## Top 10 Products

```python
top_products = data.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)
top_products.plot(kind="bar")
plt.title("Top 10 Products")
plt.show()
```

A small number of products contribute significantly to total revenue.

## Region-wise Sales

```python
region_sales = data.groupby("Region")["Sales"].sum()
region_sales.plot(kind="bar")
plt.title("Sales by Region")
plt.show()
```

Sales vary across regions, indicating differences in market demand.

##  Final Insights & Business Recommendations

### Key Insights
- Sales show growth with seasonal fluctuations  
- Top products drive majority of revenue  
- Regional differences impact performance  

### Recommendations
- Focus on high-performing products  
- Optimize sales strategies during peak periods  
- Improve performance in low-performing regions
